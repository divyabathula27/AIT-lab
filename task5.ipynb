```python
import random

path1 = "Minimum path"
path2 = "Long path"

probability_path1 = 0.70
probability_path2 = 0.30

ants = 10

path1_count = 0
path2_count = 0

for i in range(ants):
    choice = random.choices(
        [path1, path2],
        weights=[probability_path1, probability_path2]
    )[0]

    if choice == path1:
        path1_count += 1
    else:
        path2_count += 1

print("Minimum path selected:", path1_count)
print("Long path selected:", path2_count)

if path1_count > path2_count:
    print("Preferred path:", path1)
else:
    print("Preferred path:", path2)
```
