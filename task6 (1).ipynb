```python
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'D'],
    'D': ['B', 'C', 'E'],
    'E': ['D']
}
vertices = list(graph.keys())
def is_safe(vertex, color, colors):
    for neighbour in graph[vertex]:
        if colors.get(neighbour) == color:
            return False
    return True
def color_graph(index, colors, total_colors):
    if index == len(vertices):
        return True
    vertex = vertices[index]
    for color in range(1, total_colors + 1):
        if is_safe(vertex, color, colors):
            colors[vertex] = color

            if color_graph(index + 1, colors, total_colors):
                return True
            del colors[vertex]

    return False
for n in range(1, len(vertices) + 1):
    colors = {}
    if color_graph(0, colors, n):
        print("Minimum number of colors:", n)
        print("Color Assignment:")
        for vertex in vertices:
            print(vertex, "-> Color", colors[vertex])
        break
```
