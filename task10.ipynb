```python
from kanren import Relation, facts, var, run

studies = Relation()

facts(studies,
      ("Ravi", "Python"),
      ("Sita", "Python"),
      ("Ram", "Java"),
      ("Kumar", "C++"))

x = var()

python_students = run(0, x, studies(x, "Python"))
print("Students studying Python:", python_students)

subject = var()
ravi_subjects = run(0, subject, studies("Ravi", subject))
print("Subjects studied by Ravi:", ravi_subjects)
```

```python
from kanren import Relation, facts, var, run

is_animal = Relation()
has_color = Relation()

facts(is_animal, "Frog", "Crocodile", "Canary", "Flamingo")
facts(has_color,
      ("Frog", "Green"),
      ("Crocodile", "Green"),
      ("Canary", "Yellow"),
      ("Flamingo", "Pink"))

x = var()
green_animals = run(0, x, is_animal(x), has_color(x, "Green"))
print("Green animals:", green_animals)

color = var()
canary_color = run(1, color, has_color("Canary", color))
print("A canary is:", canary_color[0])
```

```python
from kanren import Relation, facts, var, run

is_fruit = Relation()
has_color = Relation()

facts(is_fruit, "Apple", "Banana", "Grapes", "Orange")

facts(has_color,
      ("Apple", "Red"),
      ("Banana", "Yellow"),
      ("Grapes", "Green"),
      ("Orange", "Orange"))

x = var()

red_fruits = run(0, x, is_fruit(x), has_color(x, "Red"))
print("Red fruits:", red_fruits)

color = var()
banana_color = run(1, color, has_color("Banana", color))
print("Color of Banana:", banana_color[0])
```
